```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create the DataFrame
df = pd.DataFrame(data={ 
    'Col1':['0','5','4','3','4','4','3','4','5','4'], 
    'Col2':['1','4','7','3','4','5','3','5','4','4'],
    'Col3':['2','5','4','3','4','4','4','3','4','5'],
    'Col4':['3','6','4','4','4','5','3','4','5','4'],
    'Col5':['4','5','4','3','4','4','3','4','5','4'],
    'Col6':['5','6','5','3','6','4','3','5','6','4'],
    'Col7':['6','5','4','3','4','4','3','4','5','4'],
    'Col8':['7','6','4','3','4','4','3','4','5','3'],
    'Col9':['8','5','4','3','4','4','3','4','5','4'],
    'Col10':['9','4','4','3','4','4','3','5','6','4']
})

# Convert data to numeric
df = df.apply(pd.to_numeric)

# Choose X and y
X = df['Col9'].values
y = df['Col7'].values

# Compute slope (m) and intercept (b) using least squares formula
m, b = np.polyfit(X, y, 1)   # degree 1 polynomial fit (linear regression)

# Predict y values
y_pred = m * X + b

# Print regression parameters
print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")

# Plot data and regression line
plt.figure(figsize=(10,5))
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Best Fit Line')
plt.title('Linear Regression (Without scikit-learn)')
plt.xlabel('Col9')
plt.ylabel('Col7')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

Definition and Purpose
The coefficient of determination (R²) is a statistical measure that indicates how well a regression model predicts or explains an outcome. It is defined as the proportion of the variance in the dependent variable that is predictable from the independent variable(s). R² values range from 0 to 1:
R² = 0: The model does not explain any of the variability of the response data around its mean.
R² = 1: The model explains all the variability of the response data around its mean.

the model has performed well R2=0.70 for example. A higher R² value indicates a better fit of the model to the data, meaning that a larger proportion of variance is explained by the model.

For example, an R² of 0.70 suggests that 70% of the variance in the dependent variable can be explained by the independent variable(s), while the remaining 30% is attributed to other factors or inherent variability.
